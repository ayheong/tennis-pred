```python
from src.features.build_features import load_matches
from src.models.bt_features import fit_bt_strengths,bt_diff
from scipy.special import expit

df = load_matches("../data/raw/atp_matches_202*.csv")

strength_map = fit_bt_strengths(df, as_of="2025-08-01")

sinner_id = int(df.loc[df["winner_name"] == "Jannik Sinner", "winner_id"].iloc[0])
alcaraz_id = int(df.loc[df["winner_name"] == "Carlos Alcaraz", "winner_id"].iloc[0])

sinner_score = strength_map[sinner_id]
alcaraz_score = strength_map[alcaraz_id]

diff_sa = bt_diff(strength_map, sinner_id,  alcaraz_id)
diff_as = bt_diff(strength_map, alcaraz_id, sinner_id)

p_sinner_beats_alcaraz = expit(diff_sa)
p_alcaraz_beats_sinner = expit(diff_as)

print("Sinner Strength Score: ", sinner_score, " Alcaraz Strength Score: ", alcaraz_score)
print("Sinner - Alcaraz: ", bt_diff(strength_map, sinner_id, alcaraz_id))
print("Alcaraz - Sinner: ", bt_diff(strength_map, alcaraz_id, sinner_id))
print("P(Sinner beats Alcaraz) ≈", p_sinner_beats_alcaraz)
print("P(Alcaraz beats Sinner) ≈", p_alcaraz_beats_sinner)
```

```python
top10_names = [
    "Jannik Sinner",
    "Carlos Alcaraz",
    "Alexander Zverev",
    "Taylor Fritz",
    "Jack Draper",
    "Ben Shelton",
    "Novak Djokovic",
    "Alex De Minaur",
    "Karen Khachanov",
    "Lorenzo Musetti",
]

print("\nTop 10 BT Strengths:")
for name in top10_names:
    pid = int(df.loc[df["winner_name"] == name, "winner_id"].iloc[0])
    score = strength_map.get(pid, 0.0)
    print(f"{name:20s} → {score:.3f}")
```
